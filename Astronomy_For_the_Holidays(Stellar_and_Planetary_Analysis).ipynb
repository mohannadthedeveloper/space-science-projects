Exoplanet Detector: Using Kepler Telescope Data


```python
import lightkurve as lk
import numpy as np

star = input("Enter the star: ")
R_s = float(input("Enter radius of host star in solar radii: "))

R_SUN = 6.9634e8
R_s_meters = R_s * R_SUN

# This gets ALL quarters and combines them
search = lk.search_targetpixelfile(star, mission="Kepler", quarter=1)
if len(search) == 0:
    print(f"ERROR: No data found for {star}")
    print("Try: 'KIC 11446443'")
    exit()

pixelFile = search[0].download()
lc = pixelFile.to_lightcurve()
periodogram = lc.to_periodogram(method='bls', 
                                period=np.linspace(0.5, 300, 20000), 
                                duration=np.linspace(0.01, 0.2, 20))

best_period = periodogram.period_at_max_power
folded = lc.fold(period=best_period.value)

flux_min = np.nanmin(folded.flux.value)
flux_median = np.nanmedian(folded.flux.value)
depth = 1 - (flux_min / flux_median)
r_p_meters = R_s_meters * np.sqrt(depth)
r_p_km = r_p_meters / 1000
r_p_jupiter = r_p_km / 71492

print("="*60)
print("MY EXOPLANET DETECTION CODE RESULTS:")
print("="*60)
print(f"Target: {star}")
print(f"Best period found: {best_period.value:.2f} days")
print(f"Transit depth: {depth:.6f}")
print(f"Planet radius: {r_p_km:.0f} km")
print(f"Planet radius: {r_p_jupiter:.3f} × Jupiter")
print("="*60)
```
